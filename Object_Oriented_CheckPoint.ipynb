```python
class Account:
  def __init__(self, firstname, lastname, address, Account_Number, Account_Holder, Account_Balance):
    self.firstname = firstname
    self.lastname = lastname
    self.address = address
    self.Account_Number = Account_Number
    self.Account_Holder = Account_Holder
    self.Account_Balance = Account_Balance


  def getinfo(self):
    return {'firstname': self.firstname,
          'lastname': self.lastname,
          'address': self.address,
          'Account_Number': self.Account_Number,
          'Account_Holder': self.Account_Holder,
          'Account_Balance': self.Account_Balance}
  def deposit(self, amount):
    self.Account_Balance = 0
    if amount > 0:
      self.Account_Balance = self.Account_Balance + amount
      print(f'Your deposit of {amount} is successful \n Total Balance is {self.Account_Balance}')

  def withdrawal(self, amount):
    if self.Account_Balance >= amount:
      self.Account_Balance = self.Account_Balance -amount
      print(f"you have withdrawn {amount} naira from your account. \n Total bal: {self.Account_Balance}")

  def Check_Balance(self, amount):
    return self.Account_Balance


my_account = Account('Samuel', 'John', 'Lagos', '0099776635', 'Samuel John', 0.00)

```
